## 1. Listas

Uma lista guarda vários valores numa variável só, entre colchetes. A posição começa em **0**.

```python
notas = [8, 6, 9]
print(notas[0])    # a primeira
print(len(notas))  # quantas são
print(sum(notas))  # a soma
```

## 2. Incluir com `append`

O `append` coloca um valor novo no fim da lista.

```python
notas = [8, 6, 9]
notas.append(7)
print(notas)
```

## ✏️ Sua vez


A lista `horas` guarda as horas de tela por dia de 7 estudantes. Um oitavo estudante respondeu depois: **8 horas**.

Antes do cálculo da média, inclua o 8 na lista e execute.

```python
horas = [2, 5, 3, 4, 6, 1, 7]
# inclua o 8 com append
media = sum(horas) / len(horas)
print("Média:", media)
```

## ✅ Confira

Execute a célula abaixo. Ela confere o que você fez; não precisa mudar nada nela.

```python
# Conferência automática
if len(horas) == 7:
    print("A lista ainda tem 7 respostas: falta incluir o 8 com append, antes da média.")
elif horas != [2, 5, 3, 4, 6, 1, 7, 8]:
    print("A lista ficou diferente do esperado:", horas, "- confira o append.")
elif media != sum(horas) / len(horas):
    print("O 8 entrou, mas a média foi calculada antes. Coloque o append antes da linha da média.")
else:
    print("✅ Certo! Código de conclusão:", media)
    print("Digite esse número no app EduCode.")
```
