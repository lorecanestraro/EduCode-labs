## 1. Criar uma função

`def` dá nome a um trecho de código. Ele só roda quando a função é chamada. Execute:

```python
def saudar(nome):
    print("Olá,", nome)

saudar("Ana")
saudar("Bia")
```

## 2. Devolver um valor

Com `return`, a função entrega o resultado a quem chamou, e ele pode entrar em outras contas.

```python
def dobro(x):
    return x * 2

print(dobro(4) + 1)
```
