## 1. Repetir com `for`

O `for` repete as linhas recuadas para cada valor do `range`.
`range(1, 6)` gera 1, 2, 3, 4 e 5: o primeiro número entra, o último não.

```python
for i in range(1, 6):
    print("Volta", i)
```

## 2. Acumular

Para somar dentro de um laço, usamos um **acumulador**: uma variável que começa em zero e cresce a cada volta.

```python
soma = 0
for i in range(1, 4):
    soma = soma + i
    print("Somei", i, "e agora a soma é", soma)
```

## ✏️ Sua vez


O programa abaixo já mostra a tabuada do 7. Faça ele **também somar** todos os resultados em `total`.

No lugar do comentário, escreva a linha que soma `n * i` ao `total`, com o mesmo recuo do `print`.

```python
n = 7
total = 0
for i in range(1, 11):
    print(n, "x", i, "=", n * i)
    # some o resultado em total
print("Soma:", total)
```

## ✅ Confira

Execute a célula abaixo. Ela confere o que você fez; não precisa mudar nada nela.

```python
# Conferência automática
esperado = sum(n * i for i in range(1, 11))
if total == 0:
    print("O total continua 0: falta a linha que acumula, dentro do for.")
elif total != esperado:
    print("Ainda não. Confira se a linha está recuada (dentro do for) e se soma n * i.")
elif n != 7:
    print("✅ A soma está certa! Volte n para 7 e execute de novo para ver o código.")
else:
    print("✅ Certo! Código de conclusão:", total)
    print("Digite esse número no app EduCode.")
```
