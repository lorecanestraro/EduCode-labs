## ✏️ Sua vez


A **prova** tem **peso 2** e o **trabalho**, **peso 1**. A média é a soma ponderada dividida por 3.

Troque o `0` pela conta da média, execute e digite **9** para a prova e **6** para o trabalho.

```python
prova = float(input("Prova: "))
trab = float(input("Trabalho: "))
# prova tem peso 2; trabalho, 1
media = 0
print("Média:", media)
```

## ✅ Confira

Execute a célula abaixo. Ela confere o que você fez; não precisa mudar nada nela.

```python
# Conferência automática
esperado = (prova * 2 + trab) / 3
if abs(media - esperado) > 1e-9:
    print("Ainda não. A prova vale o dobro, e a soma precisa acontecer antes da divisão: use parênteses.")
elif (prova, trab) != (9, 6):
    print("✅ A conta está certa! Execute a célula de cima de novo com prova 9 e trabalho 6 para ver o código.")
else:
    print("✅ Certo! Código de conclusão:", media)
    print("Digite esse número no app EduCode.")
```
