## 2. O Colab como calculadora

Sem aspas, o Python **calcula**. Com aspas, ele mostra o **texto** exatamente como foi escrito.

```python
print(5 * 50)
print("5 * 50")
```

O asterisco `*` é a multiplicação. Também valem `+`, `-` e `/` (divisão).

## ✏️ Sua vez


Uma semana tem **5 dias** de aula, com **5 aulas por dia** e **50 minutos** por aula.
Quantos minutos de aula a turma tem por semana?

Na célula abaixo, troque o `0` pela conta (sem aspas!) e execute.

```python
minutos_semana = 0  # troque o 0 pela conta
print(minutos_semana)
```

## ✅ Confira

Execute a célula abaixo. Ela confere o que você fez; não precisa mudar nada nela.

```python
# Conferência automática
if minutos_semana == 5 * 5 * 50:
    print("✅ Certo! Código de conclusão:", minutos_semana)
    print("Digite esse número no app EduCode.")
elif minutos_semana == 0:
    print("Você ainda não trocou o 0 pela conta na célula de cima.")
else:
    print("Ainda não. Multiplique os três números (dias, aulas e minutos) com *, sem aspas.")
```
