## 1. Decidir com `if`

O `if` executa as linhas de dentro só quando a condição é verdadeira.
As linhas de dentro ficam **recuadas** (4 espaços): é assim que o Python sabe o que pertence ao `if`.

```python
nota = 8
if nota >= 7:
    print("Passou!")
print("Fim")
```

## 2. Mais caminhos com `elif` e `else`

O `elif` testa outra condição; o `else` pega todo o resto. O Python executa **só o primeiro** caminho verdadeiro.
Troque o valor de `nota` e execute de novo para ver cada caminho.

```python
nota = 6
if nota >= 7:
    print("Acima da média")
elif nota >= 5:
    print("Perto da média")
else:
    print("Abaixo da média")
```
